```python
df['smoking_status'].unique()
```

```python
df['smoking_status']=df['smoking_status'].map({'Former':1, 'Never':0, 'Current':2})
```

```python
df['exercise_level'].unique()
```

```python
df['exercise_level']=df['exercise_level'].map({'Active':3, 'Light':1, 'Moderate':2, 'Sedentary':0})
```

```python
df['diet_type'].unique()
```

```python
df['diet_type']=df['diet_type'].map({'Vegetarian':1, 'Omnivore':3, 'Pescatarian':2, 'Vegan':0})
```

```python
df['sun_exposure'].unique()
```

```python
df['latitude_region'].unique()
```

```python
df['symptoms_list'].head(5)
```

```python
df=df.drop(['symptoms_list'],axis=1)
```

### Training the Model

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.drop(['has_multiple_deficiencies'],axis=1)
```

```python
y=df['has_multiple_deficiencies']
```

```python
X_train,X_test, y_train,y_test=train_test_split(X,y,train_size=0.3)
```

train model using XGboost

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6
)

model.fit(X_train, y_train)
```

```python
df['has_multiple_deficiencies'].unique()
```

Prediction

```python
y_pred = model.predict(X_test)
```

Accuracy

```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```
